```python
dpath = "/nfs/turbo/umms-indikar/shared/projects/HSC/pipeline_outputs/hsc_epi2me/"
subdirs = glob.glob(f"{dpath}/*_*")

data = []

for path in subdirs:
    group_id = path.split("/")[-1]
    print(group_id)
    json_path = os.path.join(path, f"{group_id}.config_stats.json")
    df = pd.read_json(json_path, orient='index')
    df.columns = [group_id]
    data.append(df)

df = pd.concat(data, ignore_index=False, axis=1)
print(f"{df.shape=}")
df.head()
```
