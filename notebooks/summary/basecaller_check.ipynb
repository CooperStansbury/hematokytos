```python
import pandas as pd
import pysam
```

```python
fpath = "../scrna_processing/fastq_paths.txt"
df = pd.read_csv(fpath)
df.head()
```

```python
def analyze_fastq_models(file_path):
    models = set()
    records = 0
    try:
        with pysam.FastxFile(file_path) as fastq:
            for i, record in enumerate(fastq):
                records += 1
                header = record.comment #or record.comment, or record.description, depending on your file
                header_dict = dict(pair.split("=", 1) for pair in header.split())
                models.add(header_dict['basecall_model_version_id'])

        return models, records
    except FileNotFoundError:
        print(f"File not found: {file_path}")
        return None
    except Exception as e:
        print(f"An error occurred processing {file_path}: {e}")
        return None

for _, row in df.iterrows():
    path = row['file_path']
    file_id = row['sample_id']
    models, records = analyze_fastq_models(path)
    print(f"{file_id} has with {records} reads has {models}")
    if len(models) > 1:
        print(f"\tWARNING: {file_id} has {len(models)} models")


print('done!')
```
