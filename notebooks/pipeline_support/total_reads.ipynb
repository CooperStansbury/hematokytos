```python
import json
import pandas as pd
import sys
import glob
import os
```

```python
dpath = "/nfs/turbo/umms-indikar/shared/projects/HSC/pipeline_outputs/"

file_list = glob.glob(f"{dpath}*/*/*.config*")
file_list
```

```python
result = []

for f in file_list:
    basename = os.path.basename(f)
    df = pd.read_json(f, orient='index').reset_index()
    df.columns = ['stat', 'value']
    df['basename'] = basename
    result.append(df)

result = pd.concat(result)
result
```
