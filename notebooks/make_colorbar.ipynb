```python
import matplotlib.pyplot as plt
import numpy as np

from matplotlib.colors import LinearSegmentedColormap
```

```python
def make_colorbar(
    cmap='viridis', 
    width=0.2,
    height=2.75,
    title='', 
    orientation='vertical',
    tick_labels=['low', 'high'],
):
    a = np.array([[0, 1]])
    plt.figure(figsize=(width, height), dpi=300)
    plt.imshow(a, cmap=cmap)
    plt.gca().set_visible(False)
    cax = plt.axes([0.1, 0.2, 0.8, 0.6])
    ticks = np.linspace(0, 1, len(tick_labels))
    cbar = plt.colorbar(orientation=orientation, cax=cax, label=title, ticks=ticks)
    if orientation == 'vertical':
        cbar.ax.set_yticklabels(tick_labels)
    else:
        cbar.ax.set_xticklabels(tick_labels)


def make_colorbar_h(
    cmap='viridis', 
    width=2.75,
    height=0.2,
    title='', 
    orientation='horizontal',
    tick_labels=['low', 'high'],
):
    a = np.array([[0, 1]])
    plt.figure(figsize=(width, height), dpi=300)
    plt.imshow(a, cmap=cmap)
    plt.gca().set_visible(False)
    cax = plt.axes([0.1, 0.2, 0.8, 0.6])
    ticks = np.linspace(0, 1, len(tick_labels))
    cbar = plt.colorbar(orientation=orientation, cax=cax, label=title, ticks=ticks)
    if orientation == 'vertical':
        cbar.ax.set_yticklabels(tick_labels)
    else:
        cbar.ax.set_xticklabels(tick_labels)
```

```python
cmap = LinearSegmentedColormap.from_list('test', ['w', 'orange'])
make_colorbar(cmap, title='node centrality', tick_labels=['low',  'high'])
```

```python
make_colorbar('viridis', title='expression', tick_labels=['low',  'high'])
```

```python
make_colorbar(
    'viridis',
    width=0.2,
    height=1.25,
    title='expr.',
    tick_labels=['low',  'high'])
```
